## Part C - MATPLOTLIB

### Line Plot of Patients AGE vs BP

```python
plt.figure(figsize=(8,10))
plt.plot(df["age"],df["bp"],color="red")
plt.title("Age VS Bp")
plt.xlabel("bp")
plt.ylabel("sc")
plt.show()
```

### Bar Chart Showing level of BP according to Age

```python
count=df["age"].value_counts().sort_index()
plt.bar(count.index,count.values)
plt.title("Level of Bp According to Age")
plt.xlabel("Age")
plt.ylabel("Bp")
plt.show()
```

### Histogram of Patients Age

```python
plt.hist(df["age"],bins=15,color="red",edgecolor="black")
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Freguency")
plt.show()
```

### Pie Chart Showing Patients Bacteria distribution

```python
Bacteria=df["ba"].value_counts().sort_index()
plt.pie(Bacteria,labels=["present","notpresent"],autopct="%1.1f%%")
plt.title("Bacteria Distribution")
plt.show()
```

### Scatter Plot of Patients Age VS Hemoglobin

```python
plt.scatter(df["age"],df["hemo"],color="green",edgecolor="orange",s=50)
plt.title("Age VS Hemoglobin")
plt.xlabel("Age")
plt.ylabel("Hemoglobin")
plt.show()
```

## Horizontal Bar Chart Showing CKD vs NOTCKD Patients Count

```python
patients=df["classification"].value_counts()
plt.barh(patients.index,patients.values,color="green",edgecolor="black")
plt.title("CKD VS NOT CKD")
plt.xlabel("Number of Patients")
plt.ylabel("Patient Condition")
plt.show()
```

### Box Plot of Patients Sodium

```python
plt.boxplot(df["sod"].dropna())
plt.title("Box Plot Of Sodium")
plt.ylabel("sodium")
plt.show()
```

### Multiple Line Plot Comparing Patients Blood Pressure and Serum Creatine

```python
plt.figure(figsize=(5,8))
plt.plot(df["bp"],label="bp")
plt.plot(df["sc"],label="sc")
plt.title("BLOODPRESSURE VS CREATINE")
plt.xlabel("patient ID")
plt.ylabel("value")
plt.legend()
plt.show()
```

# Part D- Seaborn

### Count Plot of Patients RBC

```python
sns.countplot(x="rbc",color="red",edgecolor="black",data=df)
plt.title("Count plot of Red Blood Cell")
plt.show()
```

### Count Plot Of Patients Bacteria

```python
sns.countplot(x="ba",color="green",edgecolor="black",data=df)
plt.title("Count Plot Of Bacteria")
plt.show()
```

### Box Plot of Patients BP by Age

```python
plt.figure(figsize=(100,50))
df.boxplot(column="bp",by="age")
plt.title("Bp by Patient Age")
plt.suptitle("")
plt.xlabel("Patient Age")
plt.ylabel("Patient BP")
plt.show()
```

### Violin Plot of Patients Sugar Level by Age

```python
sns.violinplot(x="su",y="age",data=df)
plt.title("Sugar Level by Age")
plt.xlabel("Sugar Level")
plt.ylabel("Age")
plt.show()
```

### Pair Plot of numerical columns

```python
sns.pairplot(df[["age","hemo","sc","bp"]])
plt.show()
```

### Heatmap Showing Correlation Matrix

```python
corr=df[["age","bp","sc"]].corr()
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("Correlation Matrix Heatmap")
plt.show()
```

### Scatter Plot of Patients BloodPressure VS Serum Creatine with Age

```python
sns.scatterplot(x="bp",y="sc",hue="age",data=df)
plt.title("BLOOD PRESSURE VS CREATINE")
plt.xlabel("Blood Pressure")
plt.ylabel("Serum Creatine")
plt.show()
```

### Bar Plot Showing Average Blood Glucose Level by Sugar Level

```python
sns.barplot(x="su",y="bgr",data=df)
plt.title("Average Ceatine level by bp ")
plt.xlabel("Sugar Level")
plt.ylabel("Blood Glucose Random")
plt.show()
```

### KDE Plot of Patients Age

```python
sns.kdeplot(df["age"],fill=True)
plt.title("KDE Plot of Age")
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()
```

### Join Plot of Patients Age and Bp

```python
sns.jointplot(x="age",y="bp",data=df)
plt.show()
```

### Distribution Plot of Patients Albumine
